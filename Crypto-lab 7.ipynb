```python
#2 Double Columnar Transposition Cipher

def encrypt(plaintext, key):
    plaintext = plaintext.replace(" ", "").upper()

    columns = len(key)
    rows = (len(plaintext) + columns - 1) // columns

    padded = plaintext.ljust(rows * columns, 'X')

    table = [
        padded[i:i + columns]
        for i in range(0, len(padded), columns)
    ]

    order = sorted(range(columns), key=lambda i: (key[i], i))

    ciphertext = ""

    for col in order:
        for row in table:
            ciphertext += row[col]

    return ciphertext


def decrypt(ciphertext, key, original_length):
    columns = len(key)
    rows = len(ciphertext) // columns

    table = [[''] * columns for _ in range(rows)]

    order = sorted(range(columns), key=lambda i: (key[i], i))

    index = 0

    for col in order:
        for row in range(rows):
            table[row][col] = ciphertext[index]
            index += 1

    plaintext = ""

    for row in range(rows):
        for col in range(columns):
            plaintext += table[row][col]

    return plaintext[:original_length]


# Main program

plaintext = input("Enter plaintext: ")
key1 = input("Enter first numeric key: ")
key2 = input("Enter second numeric key: ")

if not key1.isdigit() or not key2.isdigit():
    print("Invalid keys")

elif not plaintext.replace(" ", ""):
    print("Plaintext cannot be empty")

else:
    original_length = len(plaintext.replace(" ", ""))

    first_cipher = encrypt(plaintext, key1)

    final_cipher = encrypt(first_cipher, key2)

    print("\nFirst Ciphertext:", first_cipher)
    print("Final Ciphertext:", final_cipher)

    intermediate = decrypt(final_cipher, key2, len(first_cipher))

    decrypted = decrypt(intermediate, key1, original_length)

    print("\nAfter first decryption:", intermediate)
    print("Final Decrypted text:", decrypted)
```
