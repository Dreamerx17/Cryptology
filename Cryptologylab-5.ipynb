```python
# Shift Cipher

def encrypt(text, key):
    result = ""

    for ch in text:
        if ch.isupper():
            result += chr((ord(ch) - 65 + key) % 26 + 65)
        elif ch.islower():
            result += chr((ord(ch) - 97 + key) % 26 + 97)
        else:
            result += ch

    return result


def decrypt(text, key):
    return encrypt(text, -key)


text = input("Enter text: ")

if text.strip() == "":
    print("Text cannot be empty!")

else:
    try:
        key = int(input("Enter shift key: "))

        choice = input("Enter E for Encrypt or D for Decrypt: ").upper()

        if choice == "E":
            encrypted = encrypt(text, key)
            print("Encrypted:", encrypted)

            decrypted = decrypt(encrypted, key)
            print("Decrypted:", decrypted)

        elif choice == "D":
            decrypted = decrypt(text, key)
            print("Decrypted:", decrypted)

        else:
            print("Invalid choice!")

    except ValueError:
        print("Key must be an integer!")
```

```python
# Caesar Cipher

def encrypt(text):
    result = ""

    for ch in text:
        if ch.isupper():
            result += chr((ord(ch) - 65 + 3) % 26 + 65)
        elif ch.islower():
            result += chr((ord(ch) - 97 + 3) % 26 + 97)
        else:
            result += ch

    return result


def decrypt(text):
    result = ""

    for ch in text:
        if ch.isupper():
            result += chr((ord(ch) - 65 - 3) % 26 + 65)
        elif ch.islower():
            result += chr((ord(ch) - 97 - 3) % 26 + 97)
        else:
            result += ch

    return result


text = input("Enter text: ")

if text.strip() == "":
    print("Text cannot be empty!")

else:
    choice = input("Enter E for Encrypt or D for Decrypt: ").upper()

    if choice == "E":
        encrypted = encrypt(text)
        print("Encrypted:", encrypted)

        decrypted = decrypt(encrypted)
        print("Decrypted:", decrypted)

    elif choice == "D":
        decrypted = decrypt(text)
        print("Decrypted:", decrypted)

    else:
        print("Invalid choice!")
```

```python
# Atbash Cipher

def atbash(text):
    result = ""

    for ch in text:
        if ch.isupper():
            result += chr(90 - (ord(ch) - 65))
        elif ch.islower():
            result += chr(122 - (ord(ch) - 97))
        else:
            result += ch

    return result


text = input("Enter text: ")

if text.strip() == "":
    print("Text cannot be empty!")

else:
    choice = input("Enter E for Encrypt or D for Decrypt: ").upper()

    if choice == "E":
        encrypted = atbash(text)
        print("Encrypted:", encrypted)

        decrypted = atbash(encrypted)
        print("Decrypted:", decrypted)

    elif choice == "D":
        decrypted = atbash(text)
        print("Decrypted:", decrypted)

    else:
        print("Invalid choice!")
```

```python
# Affine Cipher

from math import gcd


def encrypt(text, a, b):
    result = ""

    for ch in text:
        if ch.isupper():
            x = ord(ch) - 65
            result += chr((a * x + b) % 26 + 65)

        elif ch.islower():
            x = ord(ch) - 97
            result += chr((a * x + b) % 26 + 97)

        else:
            result += ch

    return result


def decrypt(text, a, b):
    inverse = pow(a, -1, 26)
    result = ""

    for ch in text:
        if ch.isupper():
            x = ord(ch) - 65
            result += chr((inverse * (x - b)) % 26 + 65)

        elif ch.islower():
            x = ord(ch) - 97
            result += chr((inverse * (x - b)) % 26 + 97)

        else:
            result += ch

    return result


text = input("Enter text: ")

if text.strip() == "":
    print("Text cannot be empty!")

else:
    try:
        a = int(input("Enter value of a: "))
        b = int(input("Enter value of b: "))

        if gcd(a, 26) != 1:
            print("Invalid a! It must be coprime with 26.")

        elif b < 0 or b > 25:
            print("Invalid b! Enter b between 0 and 25.")

        else:
            choice = input("Enter E for Encrypt or D for Decrypt: ").upper()

            if choice == "E":
                encrypted = encrypt(text, a, b)
                print("Encrypted:", encrypted)

                decrypted = decrypt(encrypted, a, b)
                print("Decrypted:", decrypted)

            elif choice == "D":
                decrypted = decrypt(text, a, b)
                print("Decrypted:", decrypted)

            else:
                print("Invalid choice!")

    except ValueError:
        print("a and b must be integers!")
```
