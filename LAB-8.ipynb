```python

# RSA Algorithm

p = 3
q = 11

n = p * q
phi = (p - 1) * (q - 1)

e = 3


for i in range(2, phi):
    if (e * i) % phi == 1:
        d = i
        break

print("Public Key:", (e, n))
print("Private Key:", (d, n))


msg = int(input("Enter a number to encrypt: "))


encrypted = pow(msg, e, n)
print("Encrypted message:", encrypted)


decrypted = pow(encrypted, d, n)
print("Decrypted message:", decrypted)
```

```python

# Diffie-Hellman Key Exchange

p = 23
g = 5


a = 6
b = 15

A = pow(g, a, p)
B = pow(g, b, p)

print("Public values:")
print("Prime number:", p)
print("Generator:", g)

print("\nmedha's public key:", A)
print("Anonymou's public key:", B)

key1 = pow(B, a, p)
key2 = pow(A, b, p)

print("\nmedha's shared key:", key1)
print("Anonymou's shared key:", key2)

if key1 == key2:
    print("\nShared secret key established successfully!")
else:
    print("\nKey exchange failed!")
```
