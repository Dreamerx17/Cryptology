```python
#1. One-Time Pad Encryption and Decryption
text = input("Enter plaintext: ").upper()
key = input("Enter key: ").upper()

if len(text) != len(key):
    print("Key length must be equal to plaintext length")
else:
    cipher = ""

    for i in range(len(text)):
        a = ord(text[i]) - ord('A')
        b = ord(key[i]) - ord('A')
        c = (a + b) % 26
        cipher = cipher + chr(c + ord('A'))

    print("Encrypted text:", cipher)

    plain = ""

    for i in range(len(cipher)):
        a = ord(cipher[i]) - ord('A')
        b = ord(key[i]) - ord('A')
        c = (a - b) % 26
        plain = plain + chr(c + ord('A'))

    print("Decrypted text:", plain)
```

```python
#2. Brute-Force Attack on Shift Cipher
cipher = "MCI RWR WH"

for key in range(26):
    plain = ""

    for ch in cipher:
        if ch >= 'A' and ch <= 'Z':
            x = (ord(ch) - ord('A') - key) % 26
            plain = plain + chr(x + ord('A'))
        else:
            plain = plain + ch

    print("Key", key, ":", plain)
    
```

```python
#3. Rail Fence Cipher Encryption
text = input("Enter plaintext: ").upper().replace(" ", "")
rails = int(input("Enter number of rails: "))

fence = [""] * rails
row = 0
direction = 1

for ch in text:
    fence[row] = fence[row] + ch

    if row == 0:
        direction = 1
    elif row == rails - 1:
        direction = -1

    row = row + direction

cipher = ""

for i in range(rails):
    cipher = cipher + fence[i]

print("Encrypted text:", cipher)
```
